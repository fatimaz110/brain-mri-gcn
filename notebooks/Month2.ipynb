```python
import numpy as np
from skimage import exposure

def preprocess_image(path):
    image = Image.open(path)
    image = image.resize((224, 224))
    image = image.convert("L")
    image = np.array(image)
    image = image / 255.0
    image = exposure.equalize_adapthist(image)
    return image
```

```python
processed_image = preprocess_image(image_path)

print("Shape:", processed_image.shape)
print("Minimum:", processed_image.min())
print("Maximum:", processed_image.max())
```

```python
import os

os.makedirs("../processed_data/tumor", exist_ok=True)
os.makedirs("../processed_data/no_tumor", exist_ok=True)

print("Folders created!")
```

```python
processed = preprocess_image(image_path)

processed_uint8 = (processed * 255).astype(np.uint8)

Image.fromarray(processed_uint8).save(
    "../processed_data/tumor/processed_image.jpg"
)

```

```python
files = os.listdir(tumor_path)

for file in files:

    input_path = os.path.join(tumor_path, file)

    try:
        processed = preprocess_image(input_path)

        processed_uint8 = (processed * 255).astype(np.uint8)

        output_path = os.path.join(
            "../processed_data/tumor",
            file
        )

        Image.fromarray(processed_uint8).save(output_path)

    except:
        print("Could not process:", file)

print("Tumor images processed!")
```

```python
files = os.listdir(no_tumor_path)

for file in files:

    input_path = os.path.join(no_tumor_path, file)

    try:
        processed = preprocess_image(input_path)

        processed_uint8 = (processed * 255).astype(np.uint8)

        output_path = os.path.join(
            "../processed_data/no_tumor",
            file
        )

        Image.fromarray(processed_uint8).save(output_path)

    except:
        print("Could not process:", file)

print("No-tumor images processed!")
```
